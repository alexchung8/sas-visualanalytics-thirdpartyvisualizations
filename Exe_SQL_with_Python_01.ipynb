```python
import requests
import pandas as pd
from pandasql import sqldf
```

```python
file_path=r"/home_equity_py.csv"
url = r"https://support.sas.com/documentation/onlinedoc/viya/exampledatasets/hom_equity.csv"
```

```python
response = requests.get(url)
with open(file_path, 'wb') as f:
  f.write(response.content)
  print(f'file download:{file_path}')
```

```python
he_df = pd.read_csv(file_path)
```

```python
sqldf('''
select * from he_df limit 10
''')
```
